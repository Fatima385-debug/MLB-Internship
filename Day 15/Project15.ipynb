```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
import glob
```

```python
os.makedirs("output", exist_ok=True)
os.makedirs("comparison", exist_ok=True)

print("Output folders created successfully.")
```

```python
dataset_path = "input" # Update this path to your dataset location

image_paths = glob.glob(os.path.join(dataset_path, "*.jpg")) + glob.glob(os.path.join(dataset_path, "*jpg")) + glob.glob(os.path.join(dataset_path, "*.jpg")) + glob.glob(os.path.join(dataset_path, "*jpg")) + glob.glob(os.path.join(dataset_path, "*.jpg")) + glob.glob(os.path.join(dataset_path, "*jpg")) 

print(f"Total Images Found: {len(image_paths)}")
```

```python
for i, path in enumerate(image_paths[:10]):
    print(f"{i+1}. {path}")
```

```python
image = cv2.imread(image_paths[0])

image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8,8))
plt.imshow(image_rgb)
plt.title("Sample Document")
plt.axis("off")
plt.show()
```

```python
# Select first 10 images
selected_images = image_paths[:10]

print("Selected Images:")

for i, path in enumerate(selected_images):
    print(f"{i+1}. {os.path.basename(path)}")
```

```python
plt.figure(figsize=(15, 12))

for i, path in enumerate(selected_images):
    image = cv2.imread(path)
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    plt.subplot(2, 5, i + 1)
    plt.imshow(image)
    plt.title(f"Image {i+1}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```
